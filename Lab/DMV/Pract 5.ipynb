```python
#Pract5 BHAGYASHREE BANSODE
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("Retail_Sales_Data.csv")
print("First five records:")
print(df.head())
print("\nDataset Information:")
df.info()
```

```python
# Total sales by region
region_sales = df.groupby(
 "Region"
)["Sales_Amount"].sum()
print("\nTotal Sales by Region:")
print(region_sales)
```

```python
# Top-performing region
top_region = region_sales.idxmax()
top_sales = region_sales.max()
print("\nTop Performing Region:", top_region)
print("Highest Sales:", top_sales)
```

```python
# Bar chart
region_sales.plot(
 kind="bar",
 figsize=(8,5)
)
plt.title("Sales Distribution by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Region and product category
region_category = df.groupby(
 ["Region", "Product_Category"]
)["Sales_Amount"].sum()
print("\nSales by Region and Product Category:")
print(region_category)
```
