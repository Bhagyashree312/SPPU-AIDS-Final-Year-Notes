```python
#Pract4 BHAGYASHREE BANSODE
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("City_Air_Quality.csv")
print(df.head())
print(df.info())
```

```python
# AQI trend
plt.figure(figsize=(10,5))
plt.plot(df["Date"], df["AQI"], marker="o")
plt.title("AQI Trend Over Time")
plt.xlabel("Date")
plt.ylabel("AQI")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# PM2.5 trend
plt.figure(figsize=(10,5))
plt.plot(df["Date"], df["PM2.5"], marker="o")
plt.title("PM2.5 Trend Over Time")
plt.xlabel("Date")
plt.ylabel("PM2.5")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Scatter plot
plt.figure(figsize=(7,5))
plt.scatter(df["PM2.5"], df["AQI"])
plt.title("Relationship Between PM2.5 and AQI")
plt.xlabel("PM2.5")
plt.ylabel("AQI")
plt.tight_layout()
plt.show()
```

```python
# Box plot
plt.figure(figsize=(6,5))
plt.boxplot(df["AQI"].dropna())
plt.title("Distribution of AQI")
plt.ylabel("AQI")
plt.show()
```
