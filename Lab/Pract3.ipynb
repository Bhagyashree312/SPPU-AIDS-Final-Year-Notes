```python
#Pract3 BHAGYASHREE BANSODE
import pandas as pd
# Load data
df = pd.read_csv("RealEstate_Prices.csv")
print("Original Data:")
print(df.head())
```

```python
# Clean column names
df.columns = (
 df.columns
 .str.strip()
 .str.lower()
 .str.replace(" ", "_")
 .str.replace(r"[^a-zA-Z0-9_]", "", regex=True)
)
print("\nCleaned Column Names:")
print(df.columns.tolist())
```

```python
# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Remove duplicate records
df = df.drop_duplicates()
```

```python
# Example: filter properties if columns exist
if "sale_price" in df.columns:
 filtered = df[df["sale_price"] > df["sale_price"].median()]
 print("\nProperties above median price:")
 print(filtered)
```

```python
# Aggregation
if "property_type" in df.columns and "sale_price" in df.columns:
 average_price = df.groupby(
 "property_type"
 )["sale_price"].mean()
 print("\nAverage Sale Price by Property Type:")
 print(average_price)
print("\nFinal Data:")
print(df.head())
```
